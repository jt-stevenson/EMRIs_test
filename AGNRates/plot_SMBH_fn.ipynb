```python
import numpy as np
import pagn.constants as ct
import matplotlib.pyplot as plt
```

```python
def Msmbh_dummy(M_smbh):
    dN_dlogM=M_smbh
    return dN_dlogM

def Msmbh_fiducial(M_smbh, beta):
    dN_dlogM=0.005*(M_smbh/(3e6*ct.MSun))**(beta) 
    return dN_dlogM


def Msmbh_PopIII(M_smbh):
    dN_dlogM=0.01*(M_smbh/(3e6*ct.MSun))**(-0.3) 
    return dN_dlogM

def Msmbh_Phenom(M_smbh):
    dN_dlogM=0.002*(M_smbh/(3e6*ct.MSun))**(0.3) 
    return dN_dlogM

def Msmbh_ARISE(M_smbh, m_b, params):
    A=params[0]
    B=params[1]
    C=params[2]
    ms=np.log10(M_smbh)
    dN_dlogM=[]
    for m in ms:
        if  m<=m_b:
            power=A*m + C
        elif m>m_b:
            power=(A+B)*m_b + B*m +C
        # print(power)
        dN_dlogM.append((10**power))
    return dN_dlogM
```

```python
MSMBHs=np.linspace(1e4,1e7,num=10000)

z_0=(-0.8, -1.67, 3.36)
z_1=(-0.37, -1.67, 0.87)
z_2=(-0.37, -1.67, -0.32)
z_3=(-0.37, -1.67, 0)
z_4=(-1.78, -1.78, 7.9)
z_max=(-1.23, -1.99, 4.33)

fiducial=Msmbh_fiducial(MSMBHs*ct.MSun, beta=0)
popiii=Msmbh_PopIII(MSMBHs*ct.MSun)
phenom=Msmbh_Phenom(MSMBHs*ct.MSun)

arise0=Msmbh_ARISE(MSMBHs, 8.5, z_0)
# arise1=Msmbh_ARISE(MSMBHs, 8.5, z_1)
# arise2=Msmbh_ARISE(MSMBHs, 8.5, z_2)
# arise3=Msmbh_ARISE(MSMBHs, 8.5, z_3)
# arise4=Msmbh_ARISE(MSMBHs, 8.5, z_4)
arisemax=Msmbh_ARISE(MSMBHs, 8.5, z_max)


plt.plot(MSMBHs, arise0, label='Vaccaro et al, z=0', color='firebrick')
# plt.plot(MSMBHs, arise1, label='z=1')
# plt.plot(MSMBHs, arise2, label='z=2')
# plt.plot(MSMBHs, arise3, label='z=3')
# plt.plot(MSMBHs, arise4, label='z=4')
plt.plot(MSMBHs, arisemax, label='Vaccaro et al, z=4.85', color='coral')

plt.plot(MSMBHs, fiducial, linestyle='--', label='Lyu et al, fiducial', color='deepskyblue')
plt.plot(MSMBHs, popiii, linestyle='--',label='Lyu et al, popiii', color='lightseagreen')
plt.plot(MSMBHs, phenom, linestyle='--', label='Lyu et al, phenom', color='royalblue')
# plt.plot(MSMBHs, arise, label='arise')
plt.xscale('log')
plt.yscale('log')
plt.ylabel(r'$dN/d$log$M_{SMBH}$')
plt.xlabel(r'$M_{SMBH}~[M_{\odot}]$')
plt.legend()
plt.savefig('/Users/pmxks13/PhD/EMRIs_test/AGNRates/plots/SMBH_MF_Comparison.png')
plt.show()
```
